## Q1

### 1. Client Motivation

The client is **long the forward**: he pays $K$ at $T_s$ and receives the bond. This is a long-duration position with financing built into the strike.

**(i) Pre-investment hedge.** The client expects a cash inflow in one year and wants to fix today the price at which it will be invested, removing reinvestment risk with no outlay now.

**(ii) Leverage.** The contract has zero MTM at inception, so it requires only variation margin instead of full principal — cheaper duration for a leveraged or balance-sheet-constrained account.

**(iii) A view on repo.** The forward price is the spot price carried at the repo rate, so the client is effectively buying financing. This is attractive if his own funding cost is worse than the dealer's repo level, or if he expects the issue to go special.

### 2. Key Term-Sheet Parameters

| Category | Terms |
|---|---|
| Underlying | CUSIP, coupon, maturity $T_m$ |
| Size | Notional = 1,000,000 **face value** |
| Dates | Trade date, settlement date $T_s$, business-day convention |
| Price | Strike $K$ — clean or dirty must be stated |
| Delivery | Physical or cash settlement |
| Accrual | Day count (ACT/ACT), accrued interest |
| Collateral | CSA: collateral currency and rate, threshold, margin frequency |

Two matter most: whether $K$ is **clean or dirty**, which fixes the cash amount at settlement, and the **CSA collateral rate**, which determines the discount factor $df_{csa}$ in equation (2).

### 3. Settlement at $T_s$

**Physical delivery (standard):**

- **Bank A delivers** 1,000,000 face of the Treasury. Since $T_s < T_m$, this is a bond with about nine years left, not a maturing one.
- **Client pays** $1{,}000{,}000 \times K/100$ if $K$ is dirty, or $1{,}000{,}000 \times (K + AI(T_s))/100$ if $K$ is clean.

**Cash settlement:** a single net payment of $1{,}000{,}000 \times (B(T_s;T_s,T_m) - K)/100$, from Bank A if $B > K$ and from the client otherwise.

**Interim:** variation margin is exchanged daily against the MTM throughout $[0,T_s]$, so $T_s$ is the final settlement, not the only cash flow.

## Q2

### 1. Why buy the spot bond at inception

Having sold the forward, the desk is **short the bond** for delivery at $T_s$: if the market rallies, the bond it owes becomes more expensive. Buying the spot bond at $t=0$ removes this exposure.

Two features make it the natural hedge:

**(i) The hedge ratio is essentially one.** The forward price is just the spot price carried to $T_s$, so the two move nearly one-for-one.

**(ii) The hedge is static.** The payoff is linear in the bond price, so nothing needs to be rebalanced — the desk buys once, holds, and at $T_s$ the bond it owns is exactly the bond it delivers. The economics are locked at trade time instead of being left exposed to a year of market moves. What remains is financing risk, not duration risk.

### 2. Funding the position

The desk now owns a bond but has used no cash of its own. It is funded through two internal legs.

**(i) Repo desk — the secured leg.** The bond is pledged as collateral against a **term repo maturing at $T_s$**, which raises most of the purchase price at rate $R$. Matching the repo maturity to the settlement date is deliberate: rolling overnight would leave the desk exposed to funding-rate moves that the forward price has already been struck against.

**(ii) Treasury desk — the unsecured leg.** Repo is over-collateralized, so the cash raised falls short of the purchase price by the haircut. The treasury desk funds that gap, plus any variation margin the desk must post under the CSA, at the bank's internal transfer price. This leg is small but expensive, since it is unsecured.

At $T_s$ the repo unwinds, the bond comes back, and it is delivered to the client against $K$. Long cash bond, short term repo, short forward — a cash-and-carry package worth zero at inception once $K = B(0;T_s,T_m)$, which is what Q3 computes.

## Q3

### 1. Setup

Write $\tau(a,b)$ for the repo accrual factor from $a$ to $b$ (ACT/360 for USD repo), and let $c_i$ be the coupons of the 10-year Treasury paid at dates $t_i \in (0, T_s]$. All prices below are **dirty** prices; the conversion to clean is made at the end.

### 2. Hint 1 — cost of the borrowed cash

Term repo pays simple interest, so borrowing \$1 at $t=0$ against the bond obliges the desk to repay at $T_s$:

$$1 + R\,\tau(0,T_s)$$

The desk borrows $B(0;0,T_m)$ to buy the bond, so the gross debt at $T_s$ is

$$B(0;0,T_m)\big(1 + R\,\tau(0,T_s)\big)$$

### 3. Hint 2 — coupons reduce the debt

The bond sits with the repo desk as collateral, so any coupon paid in $(0,T_s]$ is received by the repo desk and passed back to the rates desk as a *manufactured payment* (equivalently, netted into the repurchase price in a sell/buy-back). Either way the cash belongs to the desk, and it uses it to pay down the repo balance.

A coupon $c_i$ received at $t_i$ therefore reduces the $T_s$ debt not by $c_i$ but by $c_i$ carried forward to $T_s$ at the repo rate:

$$c_i\big(1 + R\,\tau(t_i,T_s)\big)$$

### 4. Forward price

The desk's net debt at $T_s$, which is exactly what it must receive from the client to break even, is

$$\boxed{\;B(0;T_s,T_m) \;=\; B(0;0,T_m)\big(1 + R\,\tau(0,T_s)\big) \;-\; \sum_{t_i \in (0,T_s]} c_i\big(1 + R\,\tau(t_i,T_s)\big)\;}$$

The two terms are **cost of carry** and **coupon income**, and their difference is the net carry. For an upward-sloping situation where the coupon yield exceeds the repo rate — the normal case for Treasuries in a low-repo environment — the net carry is positive and the forward price trades *below* spot.

If the strike is quoted clean, subtract accrued interest at settlement:

$$K = B(0;T_s,T_m)^{\text{dirty}} - AI(T_s)$$

and by construction $V(0) = df_{csa}(0,T_s)\big(B(0;T_s,T_m) - K\big) = 0$, i.e. the contract is struck at par.

### 5. Why this is the unique arbitrage-free price

The three legs — long cash bond, term repo to $T_s$, short forward — replicate a zero net position at $T_s$: the repo unwinds, the bond is returned and delivered to the client, and the client's payment $K$ retires the remaining debt. Since the package requires no net cash at inception and pays nothing at $T_s$, any other $K$ would be a riskless arbitrage.

### 6. Practical qualifications

- $R$ must be the **term** repo rate to $T_s$ for *this* issue. If the bond goes special, its repo rate is below GC, carry is reduced, and the forward price rises.
- Haircut funding at the treasury desk's FTP is a genuine cost not captured above; desks add it as a spread.
- The formula uses simple interest throughout, consistent with money-market convention; no discount curve appears, since financing is entirely through repo.

## Q4

### 1. What has to be recomputed

The strike $K$ is fixed at inception and never changes, so marking the contract at any $t \in (0,T_s]$ requires only two fresh inputs in equation (2):

$$V(t) = df_{csa}(t,T_s)\big(B(t;T_s,T_m) - K\big) \times \text{Notional}$$

namely the time-$t$ forward price and the time-$t$ discount factor.

### 2. Re-striking the forward price

The carry formula of Q3 is reused, with every input refreshed to time $t$:

$$B(t;T_s,T_m) = B(t;0,T_m)\big(1 + R_t\,\tau(t,T_s)\big) - \sum_{t_i \in (t,T_s]} c_i\big(1 + R_t\,\tau(t_i,T_s)\big)$$

Three things change:

| Input | Update |
|---|---|
| $B(t;0,T_m)$ | Today's spot dirty price, from the market |
| $R_t$ | The term repo rate observed at $t$ for the **remaining** period $t \to T_s$, not the original $R$ |
| $\sum c_i$ | Only coupons not yet paid; those already past drop out of the sum |

### 3. Discounting

$df_{csa}(t,T_s)$ is read off the current SOFR OIS curve, assuming USD cash collateral paying SOFR.

This makes the **dual-curve structure** explicit:

| Purpose | Curve | Reason |
|---|---|---|
| Forwarding | Repo curve $R_t$ | Repo is the true financing cost of the replicating position |
| Discounting | CSA / OIS curve | Collateral pays SOFR |

Before 2008 a single LIBOR curve served both roles; the two must now be kept separate, and the repo–OIS basis is a genuine risk factor.

### 4. Use of the MTM

$V(t)$ is the basis for daily variation margin under the CSA — the client posts collateral when $V(t) > 0$ from Bank A's perspective, and Bank A posts when it is negative — as well as the input to daily P&L and risk reporting.



## Q5

### 1. Does the desk carry market risk?

Essentially none of the first order. Long cash bond, short term repo to $T_s$, short forward — the bond price exposure cancels, and matching the repo maturity to $T_s$ locks the financing cost at $R$. What remains is second order:

| Risk | Comment |
|---|---|
| Repo rollover | Only if financing is not maturity-matched |
| Specialness | The issue going special, or a haircut increase |
| Counterparty credit (CVA) | Client default leaves the hedge as an outright position |
| Repo–OIS basis | Forwarding on repo, discounting on OIS |

### 2. No risk, no revenue?

The desk makes nothing from market direction, but the trade is still profitable: revenue comes from the **repo spread in the quote**. The desk funds at $R$ and prices the client off a worse rate $R' > R$,

$$K = B(0;0,T_m)\big(1 + R'\tau\big) - \sum_i c_i\big(1 + R'\tau_i\big)$$

A higher carry rate raises the forward price, so the client pays more at delivery. The locked-in profit is roughly

$$\text{PV} \approx df_{csa}(0,T_s) \cdot B(0;0,T_m) \cdot (R' - R)\,\tau \cdot \text{Notional}$$

known at trade time and independent of the market thereafter.

### 3. Convincing the client

The client's own cash-and-carry would cost more than $R'$:

- **Funding.** The bank repos Treasuries near GC; most clients fund tens of basis points wider, if they can term-fund at all.
- **Balance sheet.** No principal, no balance-sheet usage, only margin.
- **Operations.** No repo rolls, collateral or settlement to manage.
- **Access.** Many clients cannot reach the term repo market directly.

The gap between $R'$ and the client's own funding cost is where both sides gain: the desk earns $R' - R$, the client still beats doing it himself.

## *Q5

### The risk-neutral statement

Take the CSA zero-coupon bond $P(t,T_s)$ as numeraire and work under the $T_s$-forward measure $\mathbb{Q}^{T_s}$:

$$V(0) = P(0,T_s)\,\mathbb{E}^{T_s}\big[B(T_s;T_s,T_m) - K\big]$$

The forward price is the spot price deflated by $P(\cdot,T_s)$, hence a martingale under this measure:

$$\mathbb{E}^{T_s}\big[B(T_s;T_s,T_m)\big] = B(0;T_s,T_m)$$

Substituting recovers equation (2), with $P(0,T_s) = df_{csa}(0,T_s)$, and $V(0)=0$ gives $K = B(0;T_s,T_m)$.

## Sensitivities

Implement the two analytic sensitivity methods so that they agree with a
bump-and-reval reference:

- `gradient_wrt_ordinate`
- `gradient_of_integrated_value_wrt_ordinate`

`bump_reval_interpolator` below is a worked bump-and-reval for the interpolated
value. Mirror its structure to fill in `bump_reval_interpolator_integrand` for
the integral, then contrast both against your analytic results.

```python
def bump_reval_interpolator(
    x : float,
    axis1 : List,
    values : List,
    interp_method : str,
    extrap_method : str,
    bump_size : Optional[float] = 1e-4):

    base_interpolator = qfCreate1DInterpolator(axis1, values, interp_method, extrap_method)
    b_value = qfInterpolate1D(x, base_interpolator)

    grad = []
    for i in range(len(values)):
        values[i] += bump_size
        this_interp = qfCreate1DInterpolator(axis1, values, interp_method, extrap_method)
        bumped_value = qfInterpolate1D(x, this_interp)
        grad.append((bumped_value - b_value) / bump_size)
        values[i] -= bump_size

    return np.array(grad)


def bump_reval_interpolator_integrand(
    x_s : float,
    x_e : float,
    axis1 : List,
    values : List,
    interp_method : str,
    extrap_method : str,
    bump_size : Optional[float] = 1e-4):

    base_interpolator = qfCreate1DInterpolator(axis1, values, interp_method, extrap_method)
    b_value = qfInterpolate1DIntegral(x_s, x_e, base_interpolator)

    grad = []
    for i in range(len(values)):
        values[i] += bump_size
        this_interp = qfCreate1DInterpolator(axis1, values, interp_method, extrap_method)
        bumped_value = qfInterpolate1DIntegral(x_s, x_e, this_interp)
        grad.append((bumped_value - b_value) / bump_size)
        values[i] -= bump_size

    return np.array(grad)
```
